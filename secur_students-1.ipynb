```python

import pandas as pd
import numpy as np

# ==========================================
# STEP 1: LOAD DATA
# ==========================================
df = pd.read_csv("C:/Users/asus/Desktop/projects/students/StudentsPerformance (2).csv")

print("Original shape:", df.shape)
print("Original columns:", df.columns.tolist())

# ==========================================
# STEP 2: SELECT EXACTLY 50 STUDENTS
# ==========================================
df = df.head(50).copy().reset_index(drop=True)

# ==========================================
# STEP 3: CLEAN COLUMN NAMES
# ==========================================
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

print("\nClean column names:")
print(df.columns.tolist())

# ==========================================
# STEP 4: REMOVE UNWANTED COLUMNS
# ==========================================
df.drop(
    columns=["race_ethnicity", "lunch"],
    errors="ignore",
    inplace=True
)

# ==========================================
# STEP 5: REMOVE DUPLICATE ROWS
# ==========================================
print("\nDuplicate rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)

# Recheck: we need 50 student records
if len(df) != 50:
    raise ValueError(
        f"After duplicate removal, only {len(df)} rows remain. "
        "Check duplicates before proceeding."
    )

# ==========================================
# STEP 6: HANDLE MISSING VALUES
# ==========================================
print("\nMissing values before cleaning:")
print(df.isnull().sum())

# Clean text columns
text_cols = [
    "gender",
    "parental_level_of_education",
    "test_preparation_course"
]

for col in text_cols:
    df[col] = df[col].astype("string").str.strip()

# Standardize text casing
df["gender"] = df["gender"].str.title()

df["parental_level_of_education"] = (
    df["parental_level_of_education"].str.title()
)

df["test_preparation_course"] = (
    df["test_preparation_course"].str.title()
)

# Convert scores to numeric
score_cols = ["math_score", "reading_score", "writing_score"]

for col in score_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Fill missing scores using the median of each subject
for col in score_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing text values with an explicit label
for col in text_cols:
    df[col] = df[col].fillna("Unknown")

print("\nMissing values after cleaning:")
print(df.isnull().sum())

# ==========================================
# STEP 7: VALIDATE SCORE RANGE
# ==========================================
for col in score_cols:
    invalid = ~df[col].between(0, 100)

    if invalid.any():
        raise ValueError(
            f"{col} contains scores outside 0-100. "
            "Review these records before proceeding."
        )

# ==========================================
# STEP 8: ADD PROJECT COLUMNS
# ==========================================
df.insert(0, "student_id", range(1, 51))

# Clearly synthetic demo data, not Kaggle data
df["student_name"] = [
    f"Student {i:02d}" for i in range(1, 51)
]

df["email"] = [
    f"student{i:02d}@example.com" for i in range(1, 51)
]

df["department_id"] = 1
df["semester"] = 7
df["academic_year"] = "2026-2027"

# Calculated column
df["average_score"] = (
    df[score_cols].mean(axis=1).round(2)
)

# Derived performance category
df["performance_status"] = np.select(
    [df["average_score"] >= 75, df["average_score"] >= 50],
    ["Good", "Average"],
    default="Needs Improvement"
)

# ==========================================
# STEP 9: CREATE STUDENTS CSV
# ==========================================
student_cols = [
    "student_id",
    "student_name",
    "email",
    "department_id",
    "semester",
    "academic_year",
    "gender",
    "parental_level_of_education",
    "test_preparation_course"
]

students = df[student_cols].copy()

# ==========================================
# STEP 10: CREATE SUBJECTS CSV
# ==========================================
subjects = pd.DataFrame({
    "subject_id": [1, 2, 3],
    "subject_name": ["Math", "Reading", "Writing"]
})

# ==========================================
# STEP 11: CREATE MARKS CSV
# ==========================================
marks = df.melt(
    id_vars=["student_id"],
    value_vars=score_cols,
    var_name="source_column",
    value_name="marks"
)

marks["subject_name"] = marks["source_column"].map({
    "math_score": "Math",
    "reading_score": "Reading",
    "writing_score": "Writing"
})

marks = marks[
    ["student_id", "subject_name", "marks"]
].copy()

marks["marks"] = marks["marks"].round().astype(int)

# ==========================================
# STEP 12: SAVE CSV FILES
# ==========================================
df.to_csv("students_50_cleaned_complete.csv", index=False)
students.to_csv("students.csv", index=False)
subjects.to_csv("subjects.csv", index=False)
marks.to_csv("marks.csv", index=False)

# ==========================================
# STEP 13: FINAL VALIDATION
# ==========================================
print("\n========== FINAL RESULTS ==========")
print("Complete dataset:", df.shape)
print("Students table:", students.shape)
print("Subjects table:", subjects.shape)
print("Marks table:", marks.shape)

print("\nUnique student IDs:", df["student_id"].nunique())
print("Missing values remaining:", int(df.isnull().sum().sum()))

print("\nSaved files:")
print("1. students_50_cleaned_complete.csv")
print("2. students.csv")
print("3. subjects.csv")
print("4. marks.csv")

print("\nPreview of cleaned data:")
print(df.head())
```

```python

import pandas as pd

# 1. Read CSV files
marks_df = pd.read_csv("marks.csv")
subjects_df = pd.read_csv("subjects.csv")

# 2. Clean column names
marks_df.columns = marks_df.columns.str.strip().str.lower()
subjects_df.columns = subjects_df.columns.str.strip().str.lower()

# 3. Check columns
print("Marks columns:", marks_df.columns.tolist())
print("Subjects columns:", subjects_df.columns.tolist())

# 4. Match subject names with subject IDs
marks_df["subject_name"] = (
    marks_df["subject_name"].astype(str).str.strip().str.lower()
)
subjects_df["subject_name"] = (
    subjects_df["subject_name"].astype(str).str.strip().str.lower()
)

# 5. Convert subject_name into subject_id
subject_map = dict(
    zip(subjects_df["subject_name"], subjects_df["subject_id"])
)
marks_df["subject_id"] = marks_df["subject_name"].map(subject_map)

# Stop if any subject name has no matching ID
if marks_df["subject_id"].isna().any():
    missing = marks_df.loc[
        marks_df["subject_id"].isna(), "subject_name"
    ].unique()
    raise ValueError(f"Subject IDs not found for: {missing}")

# 6. Add semester (change 1 if your data uses another semester)
marks_df["semester"] = 1

# 7. Keep columns compatible with SQL marks table
marks_sql = marks_df[
    ["student_id", "subject_id", "marks", "semester"]
].copy()

marks_sql["student_id"] = marks_sql["student_id"].astype(int)
marks_sql["subject_id"] = marks_sql["subject_id"].astype(int)

# 8. Save the converted CSV
marks_sql.to_csv("marks_mysql_ready.csv", index=False)

print("Converted shape:", marks_sql.shape)
print(marks_sql.head())
print("Saved as marks_mysql_ready.csv")
```
